```python
import requests

url = f"https://api.themoviedb.org/3/movie/popular?api_key=589d7d8c3bd79bac594a5a7059cfd193"
response = requests.get(url)
data = response.json()
data
```

```python
import pandas as pd

movies = pd.DataFrame(data["results"])
movies['release_date'] = pd.to_datetime(movies['release_date'], errors='coerce')
```

```python
movies.head()
```

```python
movies.columns  # columns of movies
```

```python
movies.isnull().sum()  #tells about the null values in the data
```

```python
movies = movies.dropna(subset=['backdrop_path']) #drop that particular null value
```

```python
movies.isnull().sum()
```

```python
movies = movies.drop(columns=['adult', 'backdrop_path', 'poster_path', 'video'])
```

```python
movies.head()
```

```python
#AS WE CAN SEE NOW, GENRE_ID CONSISTS OF MULTIPLE VALUES  SUCH AS [12,18], we need to seperate it for relational database
movies = movies.explode('genre_ids')
```

```python
movies = movies.rename(columns={
    'id': 'movie_id',
    'title': 'movie_title',
    'vote_average': 'avg_rating'
})
movies.head()
```

```python
# 1. Create the 'movies' table
movies_table = movies.drop(columns=['genre_ids']).drop_duplicates(subset=['movie_id'])

# 2. Create the 'movie_genres' mapping table
movie_genres_table = movies[['movie_id', 'genre_ids']].rename(columns={'genre_ids': 'genre_id'})

# Check the first few rows
print("Movies Table:")
print(movies_table.head())

print("\nMovie Genres Table:")
print(movie_genres_table.head())

```

```python
movies_table.head()
```

```python
movies_table = movies_table.drop(columns=['original_title'])
```

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine('sqlite:///movies.db')
movies_table.to_sql('movies', engine, if_exists='replace', index=False)
movie_genres_table.to_sql('movie_genres', engine, if_exists='replace', index=False)
```

```python
# Read movies table
movies_from_db = pd.read_sql('SELECT * FROM movies', engine)
movies_from_db.head()

```

```python
# Read movie_genres table
movie_genres_from_db = pd.read_sql('SELECT * FROM movie_genres', engine)
movie_genres_from_db.head()
```

```python
#FIND TOP 10 MOST POPULAR MOVIES
query = """
SELECT movie_title, popularity
FROM movies
ORDER BY popularity DESC
LIMIT 10;
"""

top_10_popular = pd.read_sql(query, engine)
top_10_popular
```

```python
#Find Average Rating by Language
query = """
SELECT original_language, AVG(avg_rating) AS avg_rating
FROM movies
GROUP BY original_language;
"""

avg_rating_by_language = pd.read_sql(query, engine)
avg_rating_by_language
```

```python
query='''
SELECT STRFTIME('%Y',release_date) AS release_year,COUNT(*) AS movies_count
FROM movies
GROUP BY release_year
ORDER BY release_year;
'''

movies_per_year = pd.read_sql(query,engine)
movies_per_year
```

```python
#GET ALL MOVIES ALONG WITH THEIR GENRE_ID
query = '''
SELECT m.movie_title,mg.genre_id
FROM movies m
JOIN movie_genres mg
ON m.movie_id = mg.movie_id;
'''

movies_with_genre = pd.read_sql(query,engine)
movies_with_genre
```

```python
#TO CONCAT ALL THE GENRE ID TO A SPECIFIC MOVIE
query='''
SELECT m.movie_title,GROUP_CONCAT(mg.genre_id) as genre_ids
from movies m
JOIN movie_genres mg
ON m.movie_id = mg.movie_id
GROUP BY movie_title;
'''

movies_with_genre = pd.read_sql(query,engine)
movies_with_genre
```
